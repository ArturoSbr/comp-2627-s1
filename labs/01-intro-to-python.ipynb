## Objects

Python is an Object-Oriented Programming language. This means that the language is structured around objects and classes, rather than just functions.

### Integers

Let's declare an Integer and save it in object `y`.

````python
# Save integer 234 in a new object named y
y = 234

# View which class it belongs to
print(type(y))
````

### Floats

Think of classes like "families" or "nationalities". A class is like a family in that all of the members that belong to it share common properties, attributes and *methods*.

For example, integers can be divided.

````python
# Divide object `y` by 2 and save the result in object `z`
z = y / 2

# View which class `z` belongs to
print(type(z))
````

A property of integers is they can be divided by other integers, and they turn into floats when you do so!

### Strings

````python
# Save string "Kachow!" in object `my_string`
my_string = 'Kachow!'

# View which class it belongs to
print(type(my_string))
````

### Iterables

An "iterable" is any object that can return its elements one at a time.

#### Lists

A list is essentially a mutable array that contains objects of potentially
different types.

````python
my_list = ['this', 'is', 'a', 'list']
my_other_list = ['this is also a list', 42, 42.0]

print(type(my_list))
````

We say a list is a mutable object because we can change its values.

````python
# Grab the **THIRD** object of `my_list` (0, 1, 2) and replace its value with 12
my_list[2] = 12
print(my_list)
````

#### Tuples

A tuple is like a list in that it contains an iterable array of objects. However, a tuple is not mutable.

````python
my_tuple = ('this', 'is', 'a', 'tuple')
print(type(my_tuple))
````

Let's check that we truly cannot modify a tuple once it's been declared.

````python
my_tuple[2] = 42
````

#### Dictionaries

Dictionaries are one of the coolest types of iterables. They are pairs of keys and values used to organize data.

````python
my_dict = {
    'colors_i_like': ['black', 'gray'],
    'favorite_number': 42,
}
print(type(my_dict))
print(my_dict.get('colors_i_like'))
````

## For-Loops

`for`-loops are iterative processes applied to iterable objects that repeat a series of steps in each pass.

````python
for letter in my_string:
    print(letter)
````

````python
for element in my_list:
    print(element)
````

We can use an iterable to generate data:

````python
# Init list where we will store values iteratively (blank slate)
my_list = []

# Iterate over 0, 1, 2, ..., 9 and append it to the list
for i in range(10):
    my_list.append(i)
print(my_list)
````

Sometimes it's useful to know both the *index* of the iteration we're currently
on, as well as the value of the iterable at a given iteration.

Instead of declaring two iterables (one for the index and one for the values),
we can use `enumerate`.

````python
for indexxxxx, char in enumerate(my_string):
    print(indexxxxx, char)
````

````python
for idx, val in enumerate(my_list):
    print(f'Iteration {idx}:')
    print(f'  - Current value: {val}')
    print(f'  - Current value multiplied by 2: {val * 2}')
````

Try to write a 10-iteration Fibonacci sequence.

* A Fibonacci sequence is the sum

````python
fibonacci = [0, 1]
for i in range(8):
    lag_1 = fibonacci[-1]
    lag_2 = fibonacci[-2]
    current = lag_1 + lag_2
    fibonacci.append(current)
print(fibonacci)
````

We can slice the array to grab the last two values: `[-2:]`

````python
fibonacci = [0, 1]
for i in range(8):
    lag_1, lag_2 = fibonacci[-2:]
    fibonacci.append(lag_1 + lag_2)
print(fibonacci)
````

## While Loops

`while` loops are a version of `for` loops that simply repeat some logic *until*
a condition is met. So instead of telling it how many iterations to repeat the
logic (`for` loop), you tell it to repeat `while` a condition remains `True`.

````python
i = 0
while i < 20:
    i += 1
    # i = i + 1
print(i)
````

## If statements

`if` statements are control flow statements used to define non-overlapping
cases.

````python
# Declare some object x
x = 21

# if isinstance(x, int):
#     print("It's an int!")  # notice the use of `"` and `'`
if x >= 70:
    print("asdfasdfasdf")
elif x > 40:
    print("x is greater than 40...")
````

There are some cool shortcuts to evaluate special conditions:

````python
x = []
if x:
    print('`x` has elements inside')
else:
    print('`x` has no elements inside')
````

````python
x = ''
if x:
    print('`x` contains text')
else:
    print('`x` does not contain text')
````

We can mix `for` loops and `if` statements together!

````python
even_numbers = []
for i in range(10):
    if i % 2 == 0:
        even_numbers.append(i)
print(even_numbers)
````

## List Comprehension

List comprehension is a compact way of writing a for-loop that is much faster than
a simple for-loop.

````python
[max(i - 5, 0) for i in range(10)]
````

````python
[i for i in range(10) if i % 2 == 0]
````

## Custom Functions

Use the `def` keyword to define a function.

````python
def greet(name: str, weekday: str):
    """Greet a person.

    Print a custom greeting for a person. The custom message includes the
    person's `name` and today's day of the week.

    Args:
        name: str
            The name of the person.
        weekday: str
            The day of the week.
    """
    print(f'Hello {name}! Today is {weekday}.')
````

## Custom Classes

A class is a family of objects. This family will spawn objects from
that class, and all spawned objects will have the same

* attributes: stored data
* methods: functions that performs actions

````python
class Student:
    def __init__(self, name: str, id: int):
        self.name = name
        self.id = id
        self.knows_python = False

    def introduce(self):
        print(f"Hello, I'm {self.name}!")
````

We can spawn a `Student` *instance* by instantiating an object from it.

````python
maria = Student('Maria', 12345)
maria.introduce()
````
